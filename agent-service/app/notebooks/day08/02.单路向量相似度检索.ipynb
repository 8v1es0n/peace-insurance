### 稠密向量相似度检索

```python
from langchain_community.embeddings import DashScopeEmbeddings
from app.core.config import settings

from pymilvus import MilvusClient

# 1.创建Milvus客户端
client = MilvusClient(uri=settings.rag.milvus_url)

# 2.定义向量模型
embeddings = DashScopeEmbeddings(
    model="qwen3.7-text-embedding"
)

# 3.定义集合名称
collection_name = "collection_name_1"

# 4.定义稠密向量检索函数
def dense_search(query:str):
    # 将用户的问题进行向量化
    embed = embeddings.embed_query(query)
    # 执行检索操作
    result = client.search(
        collection_name, data=[embed], limit=3,
        output_fields=["h2", "content"],
        search_params={"metric_type": "COSINE"},
        anns_field="dense")
    # 测试
    for hits in result:
        for hit in hits:
            print("===================")
            print(hit)

# 稠密向量的作用:语义检索
# dense_search("教育的负面作用是什么?")
dense_search("给教师的建议是什么?")
```

### 稀疏向量相似度检索

```python
# 创建Milvus客户端
client = MilvusClient(uri=settings.rag.milvus_url)

# 定义集合名称
collection_name = "collection_name_1"

# 定义稀疏向量检索函数
def sparse_search(query: str):
    result = client.search(
        collection_name, data=[query], limit=3,
        output_fields=["h2", "content"],
        search_params={"metric_type": "BM25"},
        anns_field="sparse"
    )

    # 测试
    for hits in result:
        for hit in hits:
            print("===================")
            print(hit)

# 稠密向量的作用:语义检索
dense_search("教育的负面作用是什么?")
dense_search("给教师的建议的作者是谁?")
```
