```python
from langchain.agents import create_agent
from langchain_ollama import ChatOllama

# 1.默认初始化模型
model = ChatOllama(
    model="qwen3.5:4b",
    temperature=0.7
)

# 2.创建智能体
agent = create_agent(
    model=model,
    tools=[]
)

# 3.阻塞式调用智能体
response = agent.invoke({
    "messages": [
        {"role": "user", "content": "你是什么模型?"}
    ]
})

for message in response["messages"]:
    message.pretty_print()
```
