```python
# 忽略新特性警告
import warnings
warnings.filterwarnings("ignore", message=".*v3 streaming protocol.*")

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain.agents import create_agent

# 1.加载系统环境变量
load_dotenv()
# 1.默认初始化模型
model = init_chat_model(
    # 自动推断base_url、模型供应商的名称、去环境变量中自动加载对应的API_KEY
    model="deepseek-v4-flash",
    temperature=0.7,
    # 结构化输出,必须关闭思考模式
    extra_body={
        "thinking":{
            "type":"disabled"
        }
    }
)
```

```python
# 调用智能体 - 流式输出固定范式
from langchain_core.messages import HumanMessage

# 2.创建智能体
agent = create_agent(
    model=model,
    tools=[],
    system_prompt="像派大星一样说话"
)

stream = agent.stream_events(
    {
        "messages":[
            HumanMessage(content="劝你善良")
        ]
    },
    version="v3"
)

# 5.1 从流对象中拿到消息流
for msg in stream.messages:
    # 从消息流中再拿到消息片段
    for text in msg.text:
        print(text, end="")
```
