```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
from langchain_core.tools import tool
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# 加载系统环境变量
load_dotenv()

# 定义模型
model = init_chat_model(
    model="deepseek-v4-flash"
)

# 基于装饰器属性描述的工具列表
@tool("square_root", description="计算一个小数的平方根")
def square_root(x: float) -> float:
    return x ** 0.5

# 基于函数说明文档来描述工具
@tool
def get_weather(location: str, units: str = "celsius", include_forecast: bool = False) -> str:
    """
    获取当前城市的温度和天气预报
    Args:
        location: 城市名称
        units: 温度单位
        include_forecast: 是否包含天气预报
    """
    print("获取天气的工具被调用~~~")
    temp = 22 if units == "celsius" else 72
    result = f"{location} 目前天气：{temp} degrees {units[0].upper()}"
    if include_forecast:
        result += "\n后面5天: 晴朗"
    return result

tools = [square_root, get_weather]

# 定义智能体
agent = create_agent(
    model=model,
    tools=tools
)
```

```python
# 调用算平方根智能体智能体
response = agent.invoke({
    "messages":[
        HumanMessage(content="100的平方根是多少?")
    ]
})
```

```python
# 调用查询天气智能体
response = agent.invoke(
    {
        "messages": [HumanMessage(content="北京未来5天的天气怎么样？")]
    }
)
```

```python
# 解析数据
for msg in response["messages"]:
    msg.pretty_print()
```
