```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

# 1. 准备文本内容
with open("./resources/评估.md", "r", encoding="utf-8") as f:
    md_conttent = f.read()

# 2. 准备切分器
split_on = [
    ("#", "一级标题"),
    ("##", "二级标题"),
    ("###", "三级标题")
]
spliter = MarkdownHeaderTextSplitter(
    headers_to_split_on=split_on,
    strip_headers=False  # 推荐使用False,从正文中增加标题
)

# 2. 执行切分
# 注意: 只有一个split_text方法, 返回的是list[Document]
chunks = spliter.split_text(md_conttent)

for index, chunk in enumerate(chunks):
    print(
        f"""
        第{index+1}个文档
        元数据：{chunk.metadata}
        文档内容：{chunk.page_content}
        +++++++++++++++++++++++++++++++
        """
    )
```
