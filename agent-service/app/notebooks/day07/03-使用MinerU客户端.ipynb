1. 轻量解析模式

```python
from mineru import MinerU

# 1.创建客户端
client = MinerU()

# 2.调用方法执行解析
result = client.flash_extract("./resources/sample.pdf")

# 3.拿到文本内容
print(result.markdown)

# 4.将内存保存到本地
result.save_markdown("./resources/r1.md")
```

2. 精准解析模式

```python
from mineru import MinerU
from app.core.config import settings

# 创建客户端，Precision 模式，需要Token
client = MinerU(settings.rag.mineru_token)
# 解析文件
result = client.extract("./resources/sample.pdf")

# 获取markdown
print(result.markdown)
# 输出到本地
result.save_markdown("./resources/r2.md", with_images=True)
```

3. 适用于扫描件的OCR功能

```python
from mineru import MinerU
from app.core.config import settings

# 创建客户端，Precision 模式，需要Token
client = MinerU(settings.rag.mineru_token)
# 解析文件
result = client.extract("./resources/small_ocr.pdf", ocr=True)

# 获取markdown
print(result.markdown)
# 输出到本地
result.save_markdown("./resources/r3.md")
```
