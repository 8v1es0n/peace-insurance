```python
from langgraph.constants import START, END
from operator import add
from typing import TypedDict, Annotated

from langgraph.graph import StateGraph


class CustomReducerState(TypedDict):
    count:int           # 使用默认Reducer行为
    # nodes:list[str]     # 默认行为
    nodes:Annotated[list[str], add]

# 2.定义处理状态数据的节点 - 本质上就是函数
def node_1(state: CustomReducerState):
    print(f"节点1,接收原始数据:{state}")
    return {
        "count": 1,
        "nodes": ["node_1"]
    }
def node_2(state:CustomReducerState):
    print(f"节点2,接收节点1处理之后的数据:{state}")
    return {
        "count":2,
        "nodes":["node_2"]
    }
def node_3(state:CustomReducerState):
    print(f"节点2,接收节点3处理之后的数据:{state}")
    return {
        "count":3,
        "nodes":["node_2"]
    }


# 3.定义边 - 连接节点 - 形成图
builder = StateGraph(CustomReducerState)

builder.add_node("node_1", node_1)
builder.add_node("node_2", node_2)
builder.add_node("node_3", node_3)

builder.add_edge(START, "node_1")
builder.add_edge("node_1", "node_2")
builder.add_edge("node_2", "node_3")
builder.add_edge("node_3", END)

graph = builder.compile()
```

```python
# 4.调用图
from IPython.core.display import Image

display(Image(graph.get_graph(xray=True).draw_mermaid_png()))
```
